```python
import numpy as np
import tifffile as tiff

import glob
import os
```

```python
for img_path in glob.glob("data/H&E/*.tif"):
    save_path = img_path.replace("H&E", "small_H&E")
    output_dir = "/".join(save_path.split("/")[:-1])
    os.makedirs(output_dir, exist_ok=True)
    
    img = tiff.imread(img_path)
    scaled_img = img[::4, ::4, ...]
    tiff.imwrite(save_path, scaled_img)
    print(save_path)
```
